```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Chat history container
chat_history = []

# Create widgets
output_area = widgets.Output(layout=widgets.Layout(height='300px', border='1px solid #ccc', overflow='y-scroll', padding='10px'))
text_input = widgets.TwoStateButton(label='Send') if hasattr(widgets, 'TwoStateButton') else widgets.Text(placeholder='Type your message here...', layout=widgets.Layout(width='80%'))
send_button = widgets.Button(description='Send', button_style='primary', layout=widgets.Layout(width='18%'))

if not isinstance(text_input, widgets.Text):
    text_input = widgets.Text(placeholder='Type your message here...', layout=widgets.Layout(width='80%'))

input_box = widgets.HBox([text_input, send_button])
chat_bot_ui = widgets.VBox([output_area, input_box])

def format_message(sender, message):
    color = "#e1f5fe" if sender == "You" else "#f1f1f1"
    align = "right" if sender == "You" else "left"
    margin = "margin-left: 20%;" if sender == "You" else "margin-right: 20%;"
    return f"""
    <div style='text-align: {align}; margin: 5px 0;'>
        <div style='display: inline-block; background-color: {color}; padding: 8px 12px; border-radius: 15px; {margin} max-width: 80%; text-align: left;'>
            <strong>{sender}:</strong> {message}
        </div>
    </div>
    """

def get_bot_response(user_message):
    # Simple rule-based mock response. Replace with OpenAI API call if keys are configured.
    msg = user_message.lower()
    if 'hello' in msg or 'hi' in msg:
        return "Hello! How can I help you today?"
    elif 'help' in msg:
        return "I can assist you with programming, answering questions, or just chatting!"
    elif 'name' in msg:
        return "I am your Colab Assistant Bot."
    else:
        return f"I received your message: '{user_message}'. As a demo bot, I'm here to help you interface with your code!"

def handle_send(sender):
    user_message = text_input.value.strip()
    if not user_message:
        return

    # Clear input field
    text_input.value = ''

    # Append user message
    chat_history.append(("You", user_message))

    # Generate and append bot response
    bot_response = get_bot_response(user_message)
    chat_history.append(("Bot", bot_response))

    # Update output area
    with output_area:
        clear_output(wait=True)
        for s, m in chat_history:
            display(widgets.HTML(format_message(s, m)))

send_button.on_click(handle_send)
text_input.on_submit(handle_send)

# Initialize with a welcome message
with output_area:
    display(widgets.HTML(format_message("Bot", "Hello! I am your AI assistant. How can I help you today?")))

display(chat_bot_ui)
```
